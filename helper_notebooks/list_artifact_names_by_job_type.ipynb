```python
api = wandb.Api()
project_path = "eberrigan-salk-institute-for-biological-studies/sleap-roots"

runs = api.runs(project_path)
```

```python
job_type_map = {}
for run in runs:
    jt = run.job_type or "no_job_type"
    job_type_map.setdefault(jt, []).append(run.name)

# Print out results
for jt, run_names in job_type_map.items():
    print(f"\nJob type: {jt} ({len(run_names)} runs)")
    for name in run_names[:5]:  # show just first few for each
        print(f"  - {name}")
```
