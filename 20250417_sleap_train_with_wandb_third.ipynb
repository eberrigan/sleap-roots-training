```python
from pathlib import Path
from datetime import date
import re
```

```python
import sleap_roots_training

from sleap_roots_training.train import main
```

```python
# Constants for W&B initialization
ENTITY_NAME = "eberrigan-salk-institute-for-biological-studies"
PROJECT_NAME = "sleap-roots"
EXPERIMENT_NAME = "2025-04-16_generalizable_model_aug_input_scale"  # Use a unique name for this experiment
JOB_TYPE = "train"
REGISTRY = "sleap-roots-models"
```

```python
# Constants for training
SLEAP_TRAIN_COMMAND = "sleap-train {}"
USE_EXISTING_MODEL = False
```

```python
sleap_roots_training.reset_config()
sleap_roots_training.update_config(entity_name=ENTITY_NAME, project_name=PROJECT_NAME, registry=REGISTRY, experiment_name=EXPERIMENT_NAME, job_type=JOB_TYPE)
```

```python
today = date.today().isoformat()
print(f"Today's date: {today}")

base_dir = Path("D:/SLEAP/20250415_primary_root_generalizability_experiment")
```

```python
runs_dict = {
    "sorghum_soybean_canola_pennycress_rice_arabidopsis": {
        "csv_path": (base_dir / "sorghum_soybean_canola_pennycress_rice_arabidopsis" / "train_test_splits.csv").as_posix(),
        "tags": ["multi-species", "soybean", "sorghum", "canola", "arabidopsis", "pennycress", "rice", "primary", "input-scale-0_5-1_2", today],
        "model_tags": ["multi-species", "soybean", "sorghum", "canola", "arabidopsis", "pennycress", "rice", "primary", "input-scale-0_5-1_2", today],
    }
}
```

```python
# Check tags because W&B is picky
for species, args in runs_dict.items():
    for tag_type in ["tags", "model_tags"]:
        try:
            sleap_roots_training.validate_tags(args[tag_type])
        except ValueError as e:
            print(f"[{species}] Invalid {tag_type}: {e}")
```

```python
for species, args in runs_dict.items():
    artifact_name_prefix = f"{EXPERIMENT_NAME}_{species}"

    print("\n" + "=" * 60)
    print(f"🚀 Running training for: {species}")
    print(f"📄 CSV Path: {args['csv_path']}")
    print(f"🏷️ Tags: {args['tags']}")
    print(f"🧪 Model Tags: {args['model_tags']}")
    if args.get("use_sweep"):
        print("🔁 Mode: Sweep")
    elif args.get("use_existing_model"):
        print("📦 Mode: Evaluation with existing model")
    else:
        print("🎯 Mode: Single training run")
    print("=" * 60 + "\n")

    main(
        **args,
        base_dir_override=(base_dir / species).as_posix(),
        link_to_registry=True,
        artifact_name_prefix=artifact_name_prefix
    )
```
